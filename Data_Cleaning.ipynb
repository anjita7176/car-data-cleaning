```python
# Created a copy to preserve the original dataset
car = df.copy()
```

```python
# Standardized column names to lowercase and removed extra spaces
car.columns = car.columns.str.strip().str.lower().str.replace(' ', '_')
car.head()
```

```python
# Removed leading and trailing spaces from text columns
car = car.apply(
    lambda col: col.str.strip() if col.dtype == 'object' else col
)
car.head()
```

```python
# Removed duplicate rows from the dataset
car = car.drop_duplicates().copy()

# Checked updated dataset size
car.shape
```

```python
# Rechecked missing values after duplicate removal
car.isnull().sum()
```

```python
# Checked the most frequent values in the year column
car['year'].value_counts(dropna=False).head(15)
```

```python
# Converted year values to numeric and invalid entries to NaN
car['year'] = pd.to_numeric(car['year'], errors='coerce')
```

```python
# Checked price values to identify invalid text entries
car['price'].value_counts(dropna=False).head(15)
```

```python
# Removed commas and extra spaces from price values
car['price'] = car['price'].astype(str).str.replace(',', '', regex=False).str.strip()

# Replaced "Ask For Price" with missing values
car['price'] = car['price'].replace('Ask For Price', np.nan)

# Converted price values to numeric
car['price'] = pd.to_numeric(car['price'], errors='coerce')
```

```python
# Checked values in the kilometres-driven column
car['kms_driven'].value_counts(dropna=False).head(15)
```

```python
# Removed commas, kms text and extra spaces
car['kms_driven'] = (
    car['kms_driven']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('kms', '', case=False, regex=False)
    .str.strip()
)

# Converted kilometres driven to numeric
car['kms_driven'] = pd.to_numeric(car['kms_driven'], errors='coerce')
```

```python
# Checked fuel type categories and missing values
car['fuel_type'].value_counts(dropna=False)
```

```python
# Filled missing numerical values with column medians
car['year'] = car['year'].fillna(car['year'].median())
car['price'] = car['price'].fillna(car['price'].median())
car['kms_driven'] = car['kms_driven'].fillna(car['kms_driven'].median())

# Filled missing categorical values with "Unknown"
car['fuel_type'] = car['fuel_type'].fillna('Unknown')
car['name'] = car['name'].fillna('Unknown')
car['company'] = car['company'].fillna('Unknown')
```

```python
# Kept only the first two words from each car name
car['name'] = car['name'].apply(
    lambda x: ' '.join(str(x).split()[:2]) if pd.notna(x) else x
)

# Viewed the updated car names
car['name'].head(10)
```
